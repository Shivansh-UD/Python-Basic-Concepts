````python
#Modeling a Bank Account

#define a class for the bank account
class BankAccount:
    def __init__(self, owner, balance = 0): # by default when the account is first created the balance is 0
        self.owner = owner
        self.balance = balance

    #deposit method
    def depositMoney(self, amount):
        self.balance = self.balance + amount
        print(f"{amount}, has been deposited. The new balance is {self.balance}")
    
    #withdraw method
    def withdrawMoney(self, amount):
        if amount > self.balance:
            print("This amount cannot be withdrawn")
        else:
            self.balance = self.balance - amount
            print(f"{amount}, has been withdrawed from the account. The current balance is {self.balance}")

    def getBalance(self):
        return self.balance
    
#Create an account

shivansh = BankAccount("Shivansh", 10)
print(shivansh.balance)
shivansh.depositMoney(90000000000000000000000)
shivansh.getBalance()
shivansh.withdrawMoney(350000)
shivansh.getBalance()
````

````python
# Inheritance

#Parent Class
class Gadi:
    def __init__(self, windows, doors, engineType):
        self.windows = windows
        self.doors = doors
        self.engineType = engineType
    
    def drive(self):
        print(f"the person will drive the {self.engineType} car")

g1 = Gadi(4,5, "petrol")

#Child Class
class Tesla(Gadi):
    def __init__(self, windows, doors, engineType, isSelfDriving):
        super().__init__(windows, doors, engineType)
        self.isSelfDriving = isSelfDriving

    def selfDriving(self):
        print(f"The car is {self.engineType} and it can also drive by itself: {self.isSelfDriving}")
t1 = Tesla(4,4, "electric", True)
t1.selfDriving()
t1.drive()
````

# Polymorphism

## Polymorphism (Python) — Very Short & Simple

### Meaning
**Polymorphism = same action, different behavior**

---

### Simple Idea
Different objects can use the **same method name**,  
but each object does **its own version** of that action.

Python doesn’t care *what* the object is —  
only that it knows **how to respond**.

---

### Tiny Example (minimal code)
```python
class Dog:
    def speak(self):
        return "Bark"

````python
#Base class
class Animal:
    def speak(self):
        return "sound of the animal"

#Derived class 1
class Dog(Animal):
    def speak(self):
        return "Woof Woof"

#Derived class 2
class Cat(Animal):
    def speak(self):
        return "Meow Meow"
    


dog = Dog()
cat = Cat()

print(dog.speak())
print()
print(cat.speak())
````

# Abstract Base Class (ABCs)

### Abstract Base Classes are classes that define what methods a class must have, but not how they work.
### They act like a blueprint or rulebook that other classes must follow.

## Very Similar to Abstract class in Java

````python
from abc import ABC, abstractmethod #abc means Abstract base class

#define a abstract class

class Vehcile(ABC):
    @abstractmethod
    def startEngine(self):
        pass


#derived class 1
class Car1(Vehcile):
    def startEngine(self):
        return "Car engine started"
    

class Motorcycle(Vehcile):
    def startEngine(self):
        return "Motorcycle engine Started"


#function that demonstartes Polymorphism

def startVehcile(vehcile):
    print(vehcile.startEngine())

car = Car1()
motorcycle = Motorcycle()

startVehcile(car)
startVehcile(motorcycle)
````

# Encapsulation

### Encapsulation means keeping data and the methods that use that data together, and hiding internal details from the outside.

### Like a capsule or ATM machine — you don’t see how it works inside, you just use buttons provided.

1. Public
2. Protected
3. Private 

````python
# Public Access Modifiers

class Name:
    def __init__(self, fName, lName):
        self.fName = fName   # Public variable
        self.lName = lName   # Public Variable

def fullName(human):
    print(f"Your name is: {human.fName + " " + human.lName}")

h1 = Name("Shivansh", "Gupta")
fullName(h1)
````

````python
#Private access modifiers

class Name:
    def __init__(self, fName, lName, age):
        self.__fName = fName   # Private variable  A variable can be set to private by using TWO UNDERSCORES
        self.__lName = lName   # Private Variable  A variable can be set to private by using TWO UNDERSCORES
        self.age = age         #Public variable(JUST FOR TESTING. RUN THE dir(h1) to see differenece between public and private)
def fullName(human):
    print(f"Your name is: {human._Name__fName + " " + human._Name__lName}") # we can still print the private variable like this but this is NOT A GOOD PRACTICE. SO DONT.

h1 = Name("Shivansh", "Gupta", 19)

fullName(h1)


'''
When we run the line below we will see that age is showing up but fName and lname
are showing up as the first two 2 lines such as _Name__fName, this shows that they are private variables.
That is because age is a public variable and the other two are private.
'''
dir(h1)
````

````python
# Protected access modifiers
'''
Protected acces modifiers just means that the variable cannot be used outside that specific class in which 
it was originally defined in, but it can be used in a class that is derived from that original class.
'''
class Name:
    def __init__(self, fName, lName):
        self._fName = fName   # Protected variable  A variable can be set to private by using ONE UNDERSCORE
        self._lName = lName   # Protected Variable  A variable can be set to private by using ONE UNDERSCORE

class Employee(Name):
    def __init__(self, fName, lName):
        super().__init__(fName, lName) # we do this to make sure we are inheriting from the parent class
                                             # MEMORIZE THIS SYNTAX

employee = Employee("Shivansh", "Gupta")
print(employee._fName)
````

# Magic Methods 

## Magic methods are special functions in a Python class that start and end with double underscores
- __str__

## You don’t call them directly — Python calls them behind the scenes.

## basically jab hum ek class ka object banatei hai aur fir call it and print is out, hume kuch aisa result milta hai:
- <__main__.Student object at 0x...>

## isko aise nahi print karane ke liye we overwrite the magic method to get the actual value we want printed.


````python
# basic magic methods

#implimenting normally without magic methods
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

person = Person("Shivansh", 19)
print(person)
print("-----------------------------------------")

#implimenting with magic method
class Person1:
    def __init__(self, name, age):
        self.name = name
        self.age = age

#Since we over-wrote the __str__ method we are able to now see the actual information when we print as shown below. 
    def __str__(self):
        return f"Student information: {self.name}, age: {self.age}"

person1 = Person1("Shivansh", 19)
print(person1)


````

# Operator OverLoading

````python
 # Common operator overloading merthods

'''
__add__(self, other): add two objects using the + operator.
__sub__(self, other): substract two objects using the - operator.
__mul__(self, other): multiply two objects using the * operator.
__truediv__(self, other): divide two objects using the / operator.
__eq__(self, other): check if two objects are equal using the == operator.
__lt__(self, other): check is one object is less than the other using the < operator.
'''
````

````python
# custom way to add and substract vectors by changing up how the add function works
#You redefine what + means for Vector objects -> Operator Overloading

class Vector:
    def __init__(self,x, y):
        self.x = x
        self.y = y


    '''
    self: the object on the left side of the +
    other: the object on the right side of the +
    '''
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)
    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)
    
    def __repr__(self):
        return f"Vector=>({self.x},{self.y})"
    
v1 = Vector(6, 7)
v2 = Vector(1, 2)

print(v1+v2)
print(v1-v2)
````

# Generators

## yield keyword important hai.

## What does yield do?
### yield works similarly to real life with cars, it loops through and at each iteration it waits to see if we are priniting it or doing any other operation and then once thats done, it goes to the next element. 

````python
'''
What are generators?
-> a special function that returns values one-by-one instead of returning everything at once. It uses the keyword yield instead of return
'''

# example of generator
def count_up_to(n):
    i = 1
    while i <= n:
        yield i      # pause and send value
        i += 1

gen = count_up_to(5)

print(next(gen))  # 1
print(next(gen))  # 2
print(next(gen))  # 3
````

#### Example of real world usage of Generator includes reading a large file. Example shown below.

````python
def readlargeFile(path):
    with open(path, 'r') as file:
        for line in file:
            yield line
````

````python
filePath = 'largeTextFileForTesting.txt'

for line in readlargeFile(filePath):
    print(line.strip())
````

# Decorators

### a function that modifies another function’s behavior without changing its actual code.

⭐ Basic Idea

A decorator:

takes a function

adds something before/after it

returns a new improved function

````python
# function copy

def func1():
    return "welcome"

wel = func1 #assigning the function to a variable
print(wel()) # basically using the variable to call the function
del func1 # deleting the original function
print(wel()) #calling the variable assigned earlier to see if it still works for the function
````

````python
#Closures
def mainWelcome(msg1):
    msg = "welcome to program gang"
    def subMethod():
        print("welcome to the program twin!!")
        print(msg) # inside a submethod we can also call any variables we defined in the main function
        print(msg1)
        print("learn the python stuff broo")
    
    return subMethod()

mainWelcome("hello")
````

````python
# Decorators

def my_decorator(func):
    def wrapper():
        print("Before function runs")
        func()
        print("After function runs")
    return wrapper


'''
the below code is same as the following 

say_hello = my_decorator(say_hello)

'''
@my_decorator # as soon as we add this line, it means that whenever we call the say_hello() method, the my_decorator() method will be called aswell.
def say_hello():
    print("Hello!")

say_hello()
````
